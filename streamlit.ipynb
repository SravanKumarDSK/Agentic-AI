## StreamLIt 
### an open-source Python framework used to build interactive web applications quickly, especially for Data Science, Machine Learning, and AI projects.

```python
%pip install streamlit
```

```python
import streamlit as st

st.title("My First StreamLit App Created by Sravan Kumar")

st.write("Welcome This app calculate the square of a number.")

st.header("Select a Number")

number=st.slider("Pick a number",0,100,25)

st.subheader("Result")

squared_number=number*number

st.write(f"The square of **{number}** is**{squared_number}**.")
```

```python
!python -m streamlit run app.py
```
